```python
import pymysql
```

```python
connection = pymysql.connect(host='localhost',
                             user='root',
                             password='mysql')
```

```python
cursor=connection.cursor()
```

```python
### database of hospital
query= "create database hospital"

cursor.execute(query)
connection.commit()
```

```python
### create tables in hospital db

query= """ use hospital;
create table doctors (
doctor_id int PRIMARY KEY,
  first_name varchar(50),
  last_name varchar(50),
  specialty varchar(50),
  phone_number varchar(50),
  email varchar(50)

);

create table patients(
  patient_id int PRIMARY KEY,
  first_name varchar(50),
  last_name varchar(50),
  gender char,
  date_of_birth DATE,
  phone_number VARCHAR(50),
  city VARCHAR(50),
  email VARCHAR(50)

);

create table Appointments(
appointment_id INT PRIMARY KEY,
  patient_id int,
  doctor_id int,
  appointment_date DATE,
  appointment_time TIME,
  reason VARCHAR(100),
  status VARCHAR(50));

create table Treatments(
treatment_id varchar(255) PRIMARY KEY,
  appointment_id INT,
  diagnosis VARCHAR(100),
  treatment_details VARCHAR(100),
  prescription VARCHAR(100),
  treatment_date DATE);


create table Billing (
  billing_id INT PRIMARY KEY,
  treatment_id varchar(255),
  amount DECIMAL(10,5),
  payment_status VARCHAR(50),
  billing_date DATE);

ALTER TABLE Billing ADD FOREIGN KEY (treatment_id) REFERENCES Treatments (treatment_id);

ALTER TABLE Treatments ADD FOREIGN KEY (appointment_id) REFERENCES Appointments (appointment_id);

ALTER TABLE Appointments ADD FOREIGN KEY (doctor_id) REFERENCES doctors (doctor_id);

ALTER TABLE Appointments ADD FOREIGN KEY (patient_id) REFERENCES patients (patient_id);"""

  
```

```python
for statement in query.split(";"):
    if statement.strip():
        cursor.execute(statement)
connection.commit()
```
